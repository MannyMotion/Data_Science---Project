    a.adding multiple columns in a single line of code
    b.when you need to overwrite the values of an exisiting columns (best practice)

   SO it returns a new object (a copy) with all the original columns in addition to the new ones.

Sort a DataFrame

```python
df_exams
```

```python
#sort by one column 
df_exams.sort_values(by='math score')
```

```python
#sort descending by one column
df_exams.sort_values('math score', ascending=False)#By using 'False' instead of ascending up it become desending going down. If ascending then 'True'
```

```python
#sort descending by multiple columns
df_exams.sort_values(['math score', 'reading score'], ascending=False)
```

```python
#sort desending by multiple columns and update dataframe.
df_exams.sort_values(['math score', 'reading score'], ascending=False,#same code for sorting descending
                    inplace=True)#this inplace is updating the df called exams
df_exams#show the dataframe
```

```python
#sorting descending with a key function
df_exams.sort_values('race/ethnicity', ascending=True,
                    key=lambda col:col.str.lower())
#this is saying sort the values inside the column 'race/ethnicity' in ascending order but in lowercase
```

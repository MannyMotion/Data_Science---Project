```python
def gc_content(seq):
    g = seq.count("G") #counts how many gs
    c = seq.count("C") #counts how many cs
    return (g + c) / len(seq) * 100

print(gc_content("ATGCGCGGAT")) 
print(gc_content("GGCC")) 
print(gc_content("ATAT")) 
```
